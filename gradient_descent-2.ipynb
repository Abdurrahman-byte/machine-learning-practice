```python
weight = 4
learning_rate = 0.1
x = 2
actual = 12
bias = 0

# Predicted value
prediction = weight * x + bias
print("Prediction:", prediction)

# finding gradient 
gradient = -2 * x * (actual - prediction)
print("Gradient:", gradient)

# Updated weight
new_weight = weight - learning_rate * gradient
print("New Weight:", new_weight)
```

# Gradient Decent Of A Single Datapoint


Goal: Make the model update its weight five times automatically using a for loop.

```python
weight = 4
learning_rate = 0.1
x = 2
actual = 12
bias = 0

for i in range(5):
    prediction = weight * x + bias
    gradient = -2 * x * (actual - prediction)
    new_weight = weight - learning_rate * gradient
    
print(f"Step {i + 1}: New weight = {new_weight}")    
```
