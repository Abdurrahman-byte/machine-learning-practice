# Prediction Error

```python
study_hours = [1, 2, 3, 4, 5]
exam_scores = [35, 45, 55, 65, 75]

weight = 10
bias = 25

hours = 6

prediction = (weight * hours + bias)

print(prediction)
```

```python
predictions = []

for hours in study_hours:
    prediction = (weight * hours + bias)
    predictions.append(prediction)

print("Predictions:", predictions)  

actual = 90
predicted = 84

error = actual - predicted

squared_error = error ** 2
print("Error:", error)
print("Squared error", squared_error)
```

# Mean Squared Error (MSE)

```python
squared_errors = [4, 16, 25]

total_error = sum(squared_errors)

number_of_prediction = 3

mse = (int(total_error) / int(number_of_prediction))

print("Total error:", total_error)
print("Number Of Prediction:", number_of_prediction)
print("MSE:", mse)
```
